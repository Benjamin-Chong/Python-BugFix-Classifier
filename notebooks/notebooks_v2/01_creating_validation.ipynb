# V2: Creating / Saving the Validation Set

## 1. Split Train and Validation Set

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pathlib import Path

train = pd.read_csv('data/processed/train_v1.csv')

train_v2, validation_v2 = train_test_split(train, test_size=.20, random_state=42, stratify=train['top_level_label'])

train_v2 = train_v2.reset_index(drop=True)
validation_v2 = validation_v2.reset_index(drop=True)

Path('data/processed').mkdir(parents=True, exist_ok=True)

train_v2.to_csv('data/processed/train_v2.csv', index=False)
validation_v2.to_csv('data/processed/validation_v2.csv', index=False)
```

## Conclusion

A validation set is necessary in V2 for tasks such as vocabulary evaluation, model selection, and hyperparameter tuning without using the test set during development.

The original training set was split into a new training set containing 22,809 examples and a validation set containing 5,703 examples. They were saved to `data/processed/train_v2.csv` and `data/processed/validation_v2.csv`, respectively.

Note: Stratification was used in order to preserve class distribution.
